
## Question 1: What are the five key concepts of Object-Oriented Programming (OOP)?

### Answer:
1. **Encapsulation:** Wrapping data and methods into a single unit (class) to protect the data from outside interference.  
   - **Example:**
     ```python
     class Example:
         def __init__(self):
             self.__private_var = 42  # Private variable

         def get_var(self):
             return self.__private_var  # Controlled access

     obj = Example()
     print(obj.get_var())  # Outputs: 42
     ```

2. **Abstraction:** Hiding implementation details and showing only the essential features.  
   - **Example:**
     ```python
     from abc import ABC, abstractmethod

     class Animal(ABC):
         @abstractmethod
         def sound(self):
             pass

     class Dog(Animal):
         def sound(self):
             return "Bark"

     dog = Dog()
     print(dog.sound())  # Outputs: Bark
     ```

3. **Inheritance:** Reusing code by deriving new classes from existing ones.  
   - **Example:**
     ```python
     class Parent:
         def greet(self):
             return "Hello from Parent"

     class Child(Parent):
         pass

     child = Child()
     print(child.greet())  # Outputs: Hello from Parent
     ```

4. **Polymorphism:** Allowing objects to be treated as instances of their parent class.  
   - **Example:**
     ```python
     class Shape:
         def draw(self):
             pass

     class Circle(Shape):
         def draw(self):
             return "Drawing Circle"

     class Square(Shape):
         def draw(self):
             return "Drawing Square"

     def draw_shape(shape):
         print(shape.draw())

     draw_shape(Circle())  # Outputs: Drawing Circle
     draw_shape(Square())  # Outputs: Drawing Square
     ```

5. **Association:** A relationship between objects.  
   - **Example:**
     ```python
     class Engine:
         def start(self):
             return "Engine started"

     class Car:
         def __init__(self):
             self.engine = Engine()  # Association

         def start(self):
             return self.engine.start()

     car = Car()
     print(car.start())  # Outputs: Engine started
     ```

# Types of Inheritance in Python

Inheritance in Python allows a class (child) to inherit the properties and methods of another class (parent). Below are the five types of inheritance explained with simple diagrams and examples.

## 1. Single Inheritance
A single child class inherits from one parent class.

**Diagram**:  
```plaintext
Parent
  |
Child
```

**Example**:
```python
class Parent:
    def greet(self):
        print("Hello from Parent!")

class Child(Parent):
    def greet_child(self):
        print("Hello from Child!")

obj = Child()
obj.greet()         # Inherited from Parent
obj.greet_child()   # Defined in Child
```

## 2. Multiple Inheritance
A child class inherits from more than one parent class.

**Diagram**:  
```plaintext
Parent1    Parent2
    \       /
     Child
```

**Example**:
```python
class Parent1:
    def feature1(self):
        print("Feature 1 from Parent1")

class Parent2:
    def feature2(self):
        print("Feature 2 from Parent2")

class Child(Parent1, Parent2):
    def feature3(self):
        print("Feature 3 from Child")

obj = Child()
obj.feature1()  # Inherited from Parent1
obj.feature2()  # Inherited from Parent2
obj.feature3()  # Defined in Child
```

## 3. Multilevel Inheritance
A child class inherits from a parent class, and another child class inherits from that child class (like a chain).

**Diagram**:  
```plaintext
Parent
  |
Child1
  |
Child2
```

**Example**:
```python
class Grandparent:
    def greet(self):
        print("Hello from Grandparent!")

class Parent(Grandparent):
    def greet_parent(self):
        print("Hello from Parent!")

class Child(Parent):
    def greet_child(self):
        print("Hello from Child!")

obj = Child()
obj.greet()         # Inherited from Grandparent
obj.greet_parent()  # Inherited from Parent
obj.greet_child()   # Defined in Child
```

## 4. Hierarchical Inheritance
Multiple child classes inherit from a single parent class.

**Diagram**:  
```plaintext
Parent
  /   \
Child1  Child2
```

**Example**:
```python
class Parent:
    def common_feature(self):
        print("Feature common to all children.")

class Child1(Parent):
    def feature1(self):
        print("Feature of Child1.")

class Child2(Parent):
    def feature2(self):
        print("Feature of Child2.")

obj1 = Child1()
obj2 = Child2()

obj1.common_feature()  # Inherited from Parent
obj1.feature1()        # Defined in Child1

obj2.common_feature()  # Inherited from Parent
obj2.feature2()        # Defined in Child2
```

## 5. Hybrid Inheritance
A mix of two or more types of inheritance.

**Diagram**:  
```plaintext
    Parent
    /    \
Child1   Child2
    \    /
    Child3
```

**Example**:
```python
class Parent:
    def feature(self):
        print("Feature from Parent.")

class Child1(Parent):
    def feature1(self):
        print("Feature from Child1.")

class Child2(Parent):
    def feature2(self):
        print("Feature from Child2.")

class Child3(Child1, Child2):
    def feature3(self):
        print("Feature from Child3.")

obj = Child3()
obj.feature()   # Inherited from Parent
obj.feature1()  # Inherited from Child1
obj.feature2()  # Inherited from Child2
obj.feature3()  # Defined in Child3
```

````python
#multiple inheritance
class Parent1:
    def greet(self):
        return "Hello from Parent1"

class Parent2:
    def welcome(self):
        return "Welcome from Parent2"

class Child(Parent1, Parent2):
    pass

obj = Child()
print(obj.greet())  # Outputs: Hello from Parent1
print(obj.welcome())  # Outputs: Welcome from Parent2
````

## 7.What is the Method Resolution Order (MRO) in Python? How can you retrieve it programmatically?

**MRO (Method Resolution Order)** is the order in which Python looks for a method or attribute in a hierarchy of classes. It is especially important in the context of multiple inheritance, where a method or attribute can be defined in multiple parent classes.

**MRO** ensures
1. A consistent and predictable order of method resolution.
2. Prevention of duplicate method calls in the case of "diamond inheritance" (a type of hybrid inheritance).

### retrieving MRO programmatically

#### 1. using the **`mro()`** method

````python
class A:
    def sayhi(self):
        print("hello from the class A")


class B(A):
    def sayhi(self):
        print("hello from the class B")


class C(A):
    def sayhi(self):
        print("hello from the class C")

class D(B, C):
    pass

print(D.mro())  # Output: [<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>]
````

````python
D().sayhi() # D will be searched first than the B and then C and at last A
````

#### 2. using the **`__mro__`** attribute

````python
class A:
    pass

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

print(D.__mro__)  # Output: (<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>)
````

````python
D.mro()
````

Why Use Abstract Base Classes?

   1. **Enforcing a Contract**: They ensure that subclasses provide implementations for critical methods, promoting consistency.
   2. **Polymorphism**: Abstract base classes allow you to work with objects through a common interface without worrying about their specific implementation.
   3. **Code Organization**: They help in designing systems where the structure and behavior of related classes are clear.

Key Features of Abstract Base Classes:

1.**Cannot** Be Instantiated: You cannot create an instance of an abstract class.

2.**Define** Abstract Methods: Abstract methods act as placeholders and must be implemented by derived classes.

3.**Optional** Concrete Methods: ABCs can also include methods with implementations that derived classes inherit directly.

4.**Promotes** Polymorphism: Abstract classes allow you to define common interfaces for related objects.

## 12. Question 12: Create a decorator that measures and prints the execution time of a function.

````python
import time
import math

def execution_time(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        # __name__ attribute enables getting the name of the function 
        # round function rounds the float to 2 decimals
        print(f"The function {func.__name__} took {round(end - start, 2)}seconds")
        return result
    return wrapper


  
````

````python
@execution_time
def slow_function():
    time.sleep(4) #sleep for 4 sec
    print("Function Finished executing")
    
````

## 15. Implement a static method in a class that checks if a given year is a leap year.


````python
class Ly:
    
    @staticmethod
    def isleapyear(year):
        """
        Function to check if a given year is a leap year.
        A year is a leap year if:
        - It is divisible by 4, and
        - It is not divisible by 100, unless it is also divisible by 400.
            """
        if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
            return True
        else:
            return False
````

````python
Ly.isleapyear(1900)
````
